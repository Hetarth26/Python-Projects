```python
#Program to ask the user for their name and age and print the year they will turn 100 years old

name = input("Please enter your name: ")
age = int(input("Please enter your age: "))

current_year = 2024
hundredth_year = (current_year - age) + 100 

message = name + ", you will be 100 years old in " + str (hundredth_year)
print(message)

num = int(input("Please enter a number"))
for i in range(num):
    print(message , "\n")
```
