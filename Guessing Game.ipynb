```python
import random

num = random.randint(1, 9)

guess = 0 
tries = 0 

while guess != num and guess != "exit":
    
    guess = input("Enter a guess between 1 to 9")
    if guess == "exit":
        print("Thank you for playing Guessing Game")
        break
    guess = int(guess)
    tries +=1
    if guess < num:
        print("Too low")
    elif guess > num: 
        print("Too high")
    else:
        print("Correct Guess!!")
        print("You took " + str(tries) + " attempts")
```
