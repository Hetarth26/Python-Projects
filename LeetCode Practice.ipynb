```python
#Return square root of the number given as input by the user 
import math

num = int(input("Please enter a number:  "))
if num < 0:
    print("Please enter a positive integer")
elif num > 0:
    print (int(math.sqrt(num)))
else: 
    print("ERROR")
```

```python
#Longest substring without repeating characters

s = input("Please enter a word or some characters:")
s = (set(s))
print(s)
print("The length of the string without repeating characters is" , len(s))
```

```python
#Print the palindrominc string 
s = input("Please enter a word: ")
s = s[::-1]
print(s)
```

```python
#Check whether string is palindrome 
s = input("Please enter a word: ")
if s == s[::-1]:
    print("The word is a palindrome")
else: 
    print("Then word is not a palindrome")
```

```python
#Move 0's to the end of the number 
a = [1, 2, 0, 0, 0, 3, 6]
result = [n for n in a if n != 0]
result.extend([0] * a.count(0))
print(result)
```
